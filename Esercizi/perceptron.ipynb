# Implementazione algoritmo Percettrone in Python

```python
import numpy as np

def perceptron_PR(X,Y):

    #Inizializza vettore pesi a zero
    w = np.zeros(len(X[0]))

    eta = 0.2 # Learning Rate costante
    epochs = 10 # Numero massimo di iterazioni

    for t in range(epochs):
        print(f'Strating epoch: {t}, initial w={w}')
        changes = False # Assumiamo che non ci siano cambiamenti

        # iteriamo su ogni campione nel training set
        for i in range(X.shape[0]):

            # Se il prodotto scalare tra input e pesi, moltiplicato per la label vera, è <= 0, il campione è misclassificato 
            if(np.dot(X[i],w) * Y[i]) <= 0:
                w = w + eta * X[i] * Y[i]
                print(f'Correcting w on sample: i={i}, x={X[i]}, w={w}')
                changes=True # abbiamo trovato e corretto un errore
        
        if changes is False:
            print("No other changes, solution has been found: w = ", w)
            print(f'End of training after {t} epochsm final w={w}')
            break # Nessun punto ha cambiato classe, soluzione trovata

        print(f'End of epoch: {t}, current w={w}')
```
