# Implementazione pratica di una rete neurale

Ci si affida a framework ottimizzati. **TensorFlow** è una piattaforma open-source *end-to-end* sviluppata da Google, diventata lo standard industriale per il machine learning.

All'interno di TensorFlow è integrata **Keras**, un'API di alto livello progettata specificamente per ridurre il carico cognitivo del programmatore. Keras offre interfacce coerenti e semplici, minimizza le azioni richieste per i casi d'uso comuni e fornisce messaggi di errore chiari, permettendo di tradurre la matematica in codice in pochissime righe.

Per definire l'architettura di una rete in **Keras** esistono due approcci principali: le API `Sequential` e le `Functional` API.

## Creazione di modelli con l'API Sequenziale

La classe `Sequential` è il modo più immediato per creare una rete neurale. Costruisce il modello impilando i layer uno sopra l'altro. 
Questo è l'approccio ideale quando il modello ha un singolo tensore di input e un singolo tensore di output, e procede in modo strettamente lineare da sinistra verso destra. 
La classe `Dense` viene utilizzata per implementare i layer completamente connessi. Al suo interno è possibile specificare il numero di neuroni e incorporare direttamente la funzione di attivazione desiderata. 

### Esempio di Feed Forward Network (API Sequenziale)

```python
import tensorflow as tf 
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Dense

# Inizializzazione del modello sequenziale
model = Sequential()

# Definizione della forma dell'input (un vettore di 3 elementi)
model.add(Input(shape=(3,)))

# Aggiunta dei layer nascosti e di output
model.add(Dense(3, activation="sigmoid"))
model.add(Dense(2, activation="sigmoid"))
model.add(Dense(1, activation="sigmoid"))

# Stampa un riepilogo dell'architettura
model.summary()
```

Il comando `model.summary()` è estremamente utile perché genera una tabella che mostra la forma dell'output di ogni layer e il numero di parametri da addestrare.

Nel nostro caso, il riepilogo mostrerà esattamente **23 parametri totali**. Questo conferma la formula teorica. 

### 9.9.2 Creazione di modelli complessi con Functional API

L'API Sequenziale è comoda, ma ha un limite: costringe a un'architettura rigorosamente lineare.

Se abbiamo bisogno di modelli con una **topologia non lineare** (ad esempio reti con diramazioni multiple, layer condivisi, oppure modelli con input e output multipli), dobbiamo passare alla **Keras Functional API**.

L'idea alla base dell'API Funzionale è che un modello di deep learning è, a tutti gli effetti, un **DAG (Directed Acyclic Graph - Grafo Aciclico Diretto)** di layer.

Invece di impilare strati alla cieca, si instanzia un nodo di input e lo si "passa" ai layer successivi come se fossero delle funzioni matematiche, disegnando di fatto delle frecce logiche da un layer all'altro.

**Esempio di Feed Forward Network (Functional API)**

Ricreiamo la stessa identica rete dell'esempio precedente, ma adottando il paradigma funzionale. Si noti come ogni layer venga chiamato passando come argomento l'output del layer precedente:

```python
import tensorflow as tf
from tensorflow.keras import Model
from tensorflow.keras.layers import Input, Dense

# 1. Definizione esplicita del nodo di input
inputs = Input(shape=(3,))

# 2. Chiamate funzionali: l'output del layer precedente diventa l'input del successivo
L1 = Dense(3, activation="sigmoid")(inputs)
L2 = Dense(2, activation="sigmoid")(L1)
y = Dense(1, activation="sigmoid")(L2)

# 3. Creazione del modello finale specificando nodo di partenza (inputs) e di arrivo (y)
model = Model(inputs=inputs, outputs=y, name="my_model")

# Stampa il riepilogo dell'architettura
model.summary()
```

Il risultato generato dal `summary()` sarà identico (23 parametri addestrabili), ma il modello ora è istanziato in modo molto più flessibile e pronto per eventuali modifiche architetturali avanzate.
