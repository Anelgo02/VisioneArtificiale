# Implementazione dell'algoritmo Delta Rule (alternativa al perceptron )

A livello di codice, la Delta rule è stata storicamente introdotta per addestrare il modello Adaline (Adaptive Linear Neuron) per eseguire una regressione iterativa. A differenza del percettrone, qui l'aggiornamento avviene per tutti i campioni in base all'entità dell'errore, non solo per quelli misclassificati.


```python
import numpy as np 

def adaline_delta(X,Y):
    w = np.zeros(len(X[0]))
    eta = 0.1
    epochs = 100
    t = 0
    w_old = np.ones(X[0])
    tol = 1e-4 # Tolleranza per la condizione di arresto

    while(t<epochs and np.linalg.norm(w -w_old) > tol):
        w_old = w.copy()

        # modifichiamo w per ogni campione in base all'errore
        for i in range(X.shape[0]):
            error = Y[i] - np.dot(X[i], w)
            w = w + eta * error * X[i]

        t = t+1
        print(f'End of epoch: {t}, w = {w}')
        
    if(np.linalg.norm(w - w_old) <= tol):
        print("Solution found in ", t, "epochs")
    else:
        print("Maximal number of epochs reached")
```

Se applichiamo questo apprendimento con la funzione di attivazione Sigmoide, includiamo la sua derivata nella regola di aggiornamento. Attenzione pratica: usando la sigmoide, i target $Y$ devono essere $0$ o $1$, non più $-1$ o $+1$

```python
def sigmoid(s):
    return 1.0 / (1.0 + np.exp(-s))

def adaline_delta_sigmoid(X,Y):
    Y_t = Y.copy()
    Y_t[Y_t == -1] = 0 # Adattamento label per la sigmoide

    w = np.zeros(len(X[0]))
    eta = 0.1
    epochs = 100
    t = 0
    w_old = np.ones(len(X[0]))
    tol = 1e-4

    while(t < epochs and np.linalg.norm(w - w_old) <= tol):
        w_old = w.copy()

        for i in range( X.shape[0]):
            s = np.dot(X[i],w)
            output_neuron = sigmoid(s)
            error = Y[i] - output_neuron

            # Aggiornamento pesi con la derivata della sigmoide: output * (1 - output)
            w = w + eta * error * output_neuron * (1-output_neuron) * X[i]

        t = t + 1
    
```
